**Role in the paper.** Computes the p2a method (*p2a_MS_dist*, after Michau et al. 2023) for one city
(called "method 2 / blended approach" in the file names):
- simulated urban signal: the grid cell nearest to the urban MS station among the cells with urban fraction > 50 %;
- rural reference: mean over a rural mask (within a radius of the centre, nature fraction >= 0.5,
  orography within a threshold of the city centre); the observed rural reference is CERRA averaged over the same mask
  (ANASTASIA for Rennes).

Output: `results/data/<City>/method_2_blended_approach/<City>_UHI_method2_{simulation,observations}.nc`, which
`method_3_urclimask/Eurocities_urclimmask_<City>_.ipynb` reads.

## UHI analysis 

```python
import numpy as np

# Select urban cells
urban_cells = static_input.where(static_input["sfturf"] > 50)

# Stack x/y grid into a single dimension
urban_cells = urban_cells.stack(gridcell=("y", "x"))

# Remove non-urban cells
urban_cells = urban_cells.dropna("gridcell", subset=["sfturf"])

# Calculate distance to station
dist = np.sqrt(
    (urban_cells["x"] - x_station)**2 +
    (urban_cells["y"] - y_station)**2
)

# Find closest urban cell
idx = dist.argmin()

# Extract its coordinates
x_urban = urban_cells["x"].isel(gridcell=idx).item()
y_urban = urban_cells["y"].isel(gridcell=idx).item()

print(x_urban, y_urban)

# Select the corresponding time series from da_crp
da_crp_urban = da_crp.sel(
    x=x_urban,
    y=y_urban,
    method="nearest"
)
```

```python
# Compute spatially-averaged tas within the rural mask (time series)
# Using compute_spatial_mean_over_mask function from UHI_metric module
mask_da = rural_mask['rural_mask']

# Apply the function to compute spatial mean over rural_mask for all variables in tas_crp
sim_ts = compute_spatial_mean_over_mask(
    dataset=da_crp,
    mask_da=mask_da,
    variables=['tasmin']  # None = use all data variables (tas in this case)
)
sim_ts = sim_ts.rename({'tasmin':'tasmin_sim_rur'})

sim_ts['tasmin_urb'] = da_crp_urban['tasmin']

#calculate the UHI defined as the difference between the urban point and the rural modelled temperature
sim_ts['UHI_tasmin'] = sim_ts['tasmin_urb'] - sim_ts['tasmin_sim_rur']
```

```python
da_anom = da_crp['tasmin'].mean(dim='time') - sim_ts['tasmin_sim_rur'].mean(dim='time')
da_anom
```

```python
from shapely.geometry import Polygon, MultiPolygon, box
from shapely.ops import unary_union
from matplotlib.lines import Line2D

fig, ax = plt.subplots(figsize=(10, 10),
                       subplot_kw={"projection": plot_proj})

# ------------------------------------------------------------------
# Urban fraction
# ------------------------------------------------------------------
mesh = ax.pcolormesh(
    static_input['x'],
    static_input['y'],
    (da_anom.values),
    vmin=-1.5,
    vmax=1.5,
    shading='auto',
    cmap='coolwarm',
    transform=plot_proj,
)


# ------------------------------------------------------------------
# Compute grid cell edges
# ------------------------------------------------------------------
x = static_input["x"].values
y = static_input["y"].values

dx = np.diff(x).mean()
dy = np.diff(y).mean()

x_edges = np.concatenate(([x[0]-dx/2], x[:-1]+dx/2, [x[-1]+dx/2]))
y_edges = np.concatenate(([y[0]-dy/2], y[:-1]+dy/2, [y[-1]+dy/2]))

# ------------------------------------------------------------------
# Build polygons from rural mask
# ------------------------------------------------------------------
mask = rural_mask["rural_mask"].values.astype(bool)
cells = []
ny, nx = mask.shape

for j in range(ny):
    for i in range(nx):
        if mask[j, i]:
            cells.append(
                box(
                    x_edges[i],
                    y_edges[j],
                    x_edges[i+1],
                    y_edges[j+1],
                )
            )
rural_polygon = unary_union(cells)

# ------------------------------------------------------------------
# Plot exterior + interior boundaries
# ------------------------------------------------------------------
def plot_polygon_boundaries(geom, ax, **kwargs):

    if isinstance(geom, Polygon):

        # Exterior
        x, y = geom.exterior.xy
        ax.plot(x, y, transform=plot_proj, **kwargs)

        # Interior boundaries (holes)
        for interior in geom.interiors:
            x, y = interior.xy
            ax.plot(x, y, transform=plot_proj, **kwargs)

    elif isinstance(geom, MultiPolygon):

        for poly in geom.geoms:
            plot_polygon_boundaries(poly, ax, **kwargs)

plot_polygon_boundaries(
    rural_polygon,
    ax,
    color="limegreen",
    linewidth=2,
)

# ------------------------------------------------------------------
# Station and selected grid cell
# ------------------------------------------------------------------
x_city, y_city = from_latlon_to_xy(lat_city, lon_city, plot_proj)

ax.plot(
    x_station,
    y_station,
    marker="x",
    color="red",
    markersize=7,
    transform=plot_proj,
    label=f"urban ref obs",
)

ax.plot(
    da_crp_urban["x"],
    da_crp_urban["y"],
    marker="x",
    color="darkred",
    markersize=10,
    transform=plot_proj,
    label="urban ref sim",
)

# Uncomment if desired
# ax.plot(
#     x_city,
#     y_city,
#     marker="x",
#     color="yellow",
#     markersize=10,
#     transform=plot_proj,
#     label="City centre",
# )

# ------------------------------------------------------------------
# UCDB boundary
# ------------------------------------------------------------------
ucdb_city = load_ucdb_city(
    f"{data_dir}/metadata/UCDB_shapefiles",
    city,
)

ax.add_geometries(
    ucdb_city.geometry,
    crs=ccrs.PlateCarree(),
    facecolor="none",
    edgecolor="magenta",
    linewidth=2,
)

# ------------------------------------------------------------------
# Legend
# ------------------------------------------------------------------
legend_handles = [
    Line2D([], [], color="limegreen", linewidth=2, label="Rural mask"),
    Line2D([], [], color="magenta", linewidth=2, label="UCDB boundary"),
]

handles, labels = ax.get_legend_handles_labels()
handles.extend(legend_handles)

ax.legend(handles=handles, loc="best")

# ------------------------------------------------------------------
# Cosmetics
# ------------------------------------------------------------------
ax.set_title(
    f"{city}: Illustration of p2a method"
)

ax.set_xlabel("x (m)")
ax.set_ylabel("y (m)")

plt.colorbar(
    mesh,
    ax=ax,
    shrink=0.5,
    pad=0.02,
    label="tmin anomaly (°C)",
)

plt.grid()

plt.savefig(
    f"{plot_dir}/{city}_urban_fraction_and_selected_point.png",
    dpi=200,
    bbox_inches="tight",
)

plt.show()
```
